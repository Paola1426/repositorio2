```python
numeros = [3, 15, 7, 20, 9]

mayor = numeros[0]

for n in numeros: #in dentro de
    if n > mayor:
        mayor = n

print("Mayor:", mayor)
```

```python
numero = 1
suma = 0

while numero <= 5:
    suma += numero
    numero += 1

print(suma)
```
