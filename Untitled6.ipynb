```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
x = np.linspace(0, 10, 100)
y = x**2

# Crear figura
plt.figure()

# Gráfica
plt.plot(x, y, color='blue', linestyle='-', linewidth=2, label='y = x^2')

# Etiquetas
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.title("Grafica x")  # sin acento para evitar problemas con LaTeX

# Estética
plt.legend()
plt.grid(True)

# Guardar en distintos formatos
plt.savefig("grafica.png", dpi=300, bbox_inches='tight')
plt.savefig("grafica.jpg", dpi=300, bbox_inches='tight')
plt.savefig("grafica.eps", bbox_inches='tight')

# Mostrar
plt.show()
```
